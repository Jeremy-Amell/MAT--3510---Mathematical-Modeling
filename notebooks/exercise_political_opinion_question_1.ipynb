**Specific case:** $p_X = 0.4$,\; $p_Y = 0.3$,\; $p_Z = 0.3$

where $p_X$ = Republican, $p_Y$ = Democrat, $p_Z = 1 - p_X - p_Y$ = Independent.
All analysis below uses this starting state.


## Question 1 — Find the Equilibrium Points

An equilibrium is a state $(p_X, p_Y, p_Z)$ where no net switching occurs:
$\Delta p_X = 0$ and $\Delta p_Y = 0$ simultaneously.

Using the identity $x\max(y-x,0) - y\max(x-y,0) = (x-y)\min(x,y)$,
the update equations reduce to:

$$\Delta p_X = (p_X - p_Y)\min(p_X, p_Y) + (p_X - p_Z)\min(p_X, p_Z) = 0$$

$$\Delta p_Y = (p_Y - p_X)\min(p_Y, p_X) + (p_Y - p_Z)\min(p_Y, p_Z) = 0$$

Both conditions are satisfied when one or more variables are equal or equal to zero.
Working through all cases yields **seven equilibrium points**:

| # | $(p_X,\; p_Y,\; p_Z)$ | Description |
|:-:|:---------------------:|:------------|
| 1 | $(1,\; 0,\; 0)$ | X (Republican) dominates |
| 2 | $(0,\; 1,\; 0)$ | Y (Democrat) dominates |
| 3 | $(0,\; 0,\; 1)$ | Z (Independent) dominates |
| 4 | $(\tfrac{1}{3},\; \tfrac{1}{3},\; \tfrac{1}{3})$ | Three-way equal split |
| 5 | $(\tfrac{1}{2},\; \tfrac{1}{2},\; 0)$ | X–Y two-party tie |
| 6 | $(\tfrac{1}{2},\; 0,\; \tfrac{1}{2})$ | X–Z two-party tie |
| 7 | $(0,\; \tfrac{1}{2},\; \tfrac{1}{2})$ | Y–Z two-party tie |

Starting from the specific case $(p_X, p_Y, p_Z) = (0.4, 0.3, 0.3)$,
$p_X$ is the largest, so the system converges to **equilibrium 1: X wins**.

```python
def step(px, py):
    """One discrete time step.  pz = 1 - px - py is derived."""
    pz = 1 - px - py
    xy = px * max(py - px, 0)   # X -> Y
    xz = px * max(pz - px, 0)   # X -> Z
    yx = py * max(px - py, 0)   # Y -> X
    yz = py * max(pz - py, 0)   # Y -> Z
    zx = pz * max(px - pz, 0)   # Z -> X
    zy = pz * max(py - pz, 0)   # Z -> Y
    return px + (-xy - xz + yx + zx), py + (-yx - yz + xy + zy)

# ── Specific case: simulate to confirm convergence ──
px, py = 0.4, 0.3
for t in range(200):
    px, py = step(px, py)
print(f'After 200 steps: px={px:.6f}, py={py:.6f}, pz={1-px-py:.6f}')
print('=> converges to X (Republican) wins')
```

```python
# ── Verify delta = 0 at all seven equilibria ──
equilibria = [
    (1.0,  0.0,  '(1,    0,    0)   X wins'),
    (0.0,  1.0,  '(0,    1,    0)   Y wins'),
    (0.0,  0.0,  '(0,    0,    1)   Z wins'),
    (1/3,  1/3,  '(1/3, 1/3, 1/3)  Equal split'),
    (0.5,  0.5,  '(1/2, 1/2,  0)   X-Y tie'),
    (0.5,  0.0,  '(1/2,  0,  1/2)  X-Z tie'),
    (0.0,  0.5,  '(0,   1/2, 1/2)  Y-Z tie'),
]

print(f"{'Equilibrium':<32}  dpx          dpy")
print('-' * 60)
for px0, py0, name in equilibria:
    nx, ny = step(px0, py0)
    print(f"{name:<32}  {nx-px0:+.2e}   {ny-py0:+.2e}")
```
