The specific case $(p_X, p_Y, p_Z) = (0.4, 0.3, 0.3)$ converges to

````python
from pylab import *

Js = [
    ([[0., 0.], [0., 0.]],    '(1,   0,   0)   X wins  [specific case]'),
    ([[0., 0.], [0., 0.]],    '(0,   1,   0)   Y wins'),
    ([[0., 0.], [0., 0.]],    '(0,   0,   1)   Z wins'),
    ([[2., 0.], [0., 2.]],    '(1/3,1/3, 1/3)  Equal split'),
    ([[ 1.,-1.],[-1., 1.]],   '(1/2,1/2,  0)   X-Y tie'),
    ([[ 2., 1.],[ 0., 0.]],   '(1/2, 0,  1/2)  X-Z tie'),
    ([[ 0., 0.],[ 1., 2.]],   '(0,  1/2, 1/2)  Y-Z tie'),
]

for J, name in Js:
    vals, _ = eig(J)
    print(f'Equilibrium {name}:  lambda = {[round(v.real,4) for v in vals]}')
````
