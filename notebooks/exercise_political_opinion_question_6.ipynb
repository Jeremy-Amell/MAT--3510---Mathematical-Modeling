**Specific case:** $p_X = 0.4$,\; $p_Y = 0.3$,\; $p_Z = 0.3$

## Question 6 — One Major Cause of Change in Convergence State

In this model the outcome is determined entirely by the **initial values**
$(p_X, p_Y, p_Z)$ — whichever option starts highest wins.
The specific case $(0.4, 0.3, 0.3)$ gives $p_X > p_Y = p_Z$, so X wins.
If those starting values shifted to, say, $(0.28, 0.38, 0.34)$, Y would win instead.

### Cause: A shift in initial popularity due to external events

Any event that changes the initial distribution at the **start of a new cycle**
can reverse the outcome:

- **Economic conditions** — a recession shifts $p_X$ vs.\ $p_Y$ before switching
  dynamics begin.
- **Scandals or policy failures** — reduce one option's initial popularity,
  handing the plurality to the other.
- **Candidate quality** — a strong or weak candidate changes the starting
  distribution before the feedback loop takes over.

Since the basin boundary between X winning and Y winning is the line $p_X = p_Y$
(holding $p_Z$ fixed), even a **tiny shift** across that line reverses the winner.
