# Exercise 8.1 — Bifurcation Analysis

**Textbook:** Sayama, *Introduction to the Modeling and Analysis of Complex Systems*, Section 8

---

## System

$$\frac{dx}{dt} = rx(x + 1) - x \qquad (8.17)$$

Find the critical threshold of $r$ at which a bifurcation occurs, draw a bifurcation diagram, and classify the type of bifurcation.

---

## Step 1 — Find the Equilibria

Factor the right-hand side:

$$\frac{dx}{dt} = rx^2 + rx - x = x\bigl[rx + (r - 1)\bigr]$$

Setting equal to zero gives two equilibria:

$$x^* = 0 \qquad \text{or} \qquad rx + (r-1) = 0 \;\Rightarrow\; x^* = \frac{1-r}{r}$$

The second equilibrium only coincides with the first when $\frac{1-r}{r} = 0$, i.e., when $r = 1$.

$$\boxed{r_c = 1 \quad \text{(critical threshold)}}$$

---

## Step 2 — Stability Analysis

Let $f(x) = rx^2 + (r-1)x$. The derivative is:

$$f'(x) = 2rx + (r - 1)$$

**At $x^* = 0$:**

$$f'(0) = r - 1$$

- $r < 1$: $f'(0) < 0$ → **stable**
- $r > 1$: $f'(0) > 0$ → **unstable**

**At $x^* = \dfrac{1-r}{r}$:**

$$f'\!\left(\tfrac{1-r}{r}\right) = 2r \cdot \frac{1-r}{r} + (r-1) = 2(1-r) + (r-1) = 1 - r$$

- $r < 1$: $f' = 1 - r > 0$ → **unstable** (and $x^* > 0$)
- $r > 1$: $f' = 1 - r < 0$ → **stable** (and $x^* < 0$)

**Summary:**

| $r$ | $x^* = 0$ | $x^* = (1-r)/r$ |
|:---:|:---:|:---:|
| $r < 1$ | Stable | Unstable, $x^* > 0$ |
| $r = 1$ | Both merge at $x = 0$ | — |
| $r > 1$ | Unstable | Stable, $x^* < 0$ |

The two equilibria **swap stability** at $r = 1$ without either appearing or disappearing. This is the defining signature of a **transcritical bifurcation**.

---

## Step 3 — Classification

$$\boxed{\text{Transcritical Bifurcation at } r_c = 1}$$

In a transcritical bifurcation, two equilibrium branches cross and exchange stability. Unlike a saddle-node bifurcation, neither equilibrium is created or destroyed — they simply pass through each other at the critical point and swap their stability roles.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
```

```python
# --- Bifurcation diagram ---

r_vals = np.linspace(0.1, 2.5, 500)

# Equilibrium branches
x_branch1 = np.zeros_like(r_vals)           # x* = 0
x_branch2 = (1 - r_vals) / r_vals           # x* = (1-r)/r

# Stability: branch1 stable when r<1, branch2 stable when r>1
fig, ax = plt.subplots(figsize=(9, 6))

# x* = 0: stable for r<1 (solid), unstable for r>1 (dashed)
mask_s1 = r_vals <= 1
mask_u1 = r_vals >= 1
ax.plot(r_vals[mask_s1], x_branch1[mask_s1], color='steelblue',  linewidth=2.5,
        linestyle='-',  label='$x^*=0$ (stable)')
ax.plot(r_vals[mask_u1], x_branch1[mask_u1], color='steelblue',  linewidth=2.5,
        linestyle='--', label='$x^*=0$ (unstable)')

# x* = (1-r)/r: unstable for r<1 (dashed), stable for r>1 (solid)
ax.plot(r_vals[mask_s1], x_branch2[mask_s1], color='firebrick', linewidth=2.5,
        linestyle='--', label=r'$x^*=\frac{1-r}{r}$ (unstable)')
ax.plot(r_vals[mask_u1], x_branch2[mask_u1], color='firebrick', linewidth=2.5,
        linestyle='-',  label=r'$x^*=\frac{1-r}{r}$ (stable)')

# Bifurcation point
ax.plot(1, 0, 'ko', markersize=10, zorder=5, label='Bifurcation point $(r_c=1)$')
ax.axvline(1, color='gray', linewidth=1, linestyle=':', alpha=0.7)

ax.set_xlabel('Parameter $r$', fontsize=13)
ax.set_ylabel('Equilibrium $x^*$', fontsize=13)
ax.set_title('Bifurcation Diagram — Transcritical Bifurcation at $r_c = 1$\n'
             r'$\frac{dx}{dt} = rx(x+1) - x$', fontsize=12)
ax.legend(fontsize=10, loc='upper right')
ax.axhline(0, color='black', linewidth=0.6)
ax.set_ylim(-3, 3)
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# --- 1-D phase portraits for r < 1, r = 1, r > 1 ---

def f(x, r):
    return r * x * (x + 1) - x

x_range = np.linspace(-2.5, 2.5, 400)
r_cases = [0.5, 1.0, 1.8]
titles  = ['$r = 0.5 < r_c$', '$r = 1.0 = r_c$', '$r = 1.8 > r_c$']

fig, axes = plt.subplots(1, 3, figsize=(13, 4), sharey=True)

for ax, r, title in zip(axes, r_cases, titles):
    y_vals = f(x_range, r)
    ax.plot(x_range, y_vals, color='steelblue', linewidth=2)
    ax.axhline(0, color='black', linewidth=0.8)
    ax.axvline(0, color='gray', linewidth=0.5, linestyle=':')

    # Mark equilibria and stability arrows
    eq1 = 0
    eq2 = (1 - r) / r if r != 0 else None

    for eq in [eq1, eq2]:
        if eq is None:
            continue
        stab = f(eq + 0.01, r) < 0  # True if stable
        color = 'seagreen' if stab else 'firebrick'
        marker = 'o' if stab else 'o'
        ax.plot(eq, 0, marker, color=color, markersize=10, zorder=5,
                markerfacecolor=color if stab else 'white',
                markeredgecolor=color, markeredgewidth=2)

    ax.set_title(title, fontsize=12)
    ax.set_xlabel('$x$', fontsize=12)
    ax.set_ylim(-4, 4)
    ax.set_xlim(-2.5, 2.5)
    ax.grid(alpha=0.3)

axes[0].set_ylabel('$f(x) = rx(x+1) - x$', fontsize=11)
fig.suptitle('1-D Phase Portraits  (filled = stable, open = unstable)', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# --- Time series showing stability exchange ---

def system(t, state, r):
    x = state[0]
    return [r * x * (x + 1) - x]

t_span = (0, 10)
t_eval = np.linspace(0, 10, 1000)
ics = [-0.5, -0.1, 0.1, 0.5, 1.0]

fig, axes = plt.subplots(1, 3, figsize=(13, 4), sharey=True)

for ax, r, title in zip(axes, r_cases, titles):
    colors = plt.cm.coolwarm(np.linspace(0.1, 0.9, len(ics)))
    for x0, color in zip(ics, colors):
        sol = solve_ivp(system, t_span, [x0], args=(r,), t_eval=t_eval, rtol=1e-9)
        ax.plot(sol.t, np.clip(sol.y[0], -3, 3), color=color,
                linewidth=1.6, label=f'$x_0={x0}$')
    ax.axhline(0, color='steelblue', linewidth=1, linestyle='--', alpha=0.6)
    eq2 = (1 - r) / r
    ax.axhline(eq2, color='firebrick', linewidth=1, linestyle='--', alpha=0.6)
    ax.set_title(title, fontsize=12)
    ax.set_xlabel('$t$', fontsize=12)
    ax.set_ylim(-2.5, 2.5)
    ax.grid(alpha=0.3)

axes[0].set_ylabel('$x(t)$', fontsize=12)
axes[2].legend(fontsize=9, loc='upper right')
fig.suptitle('Time Series for Various Initial Conditions', fontsize=12)
plt.tight_layout()
plt.show()
```

## Summary

| | |
|---|---|
| **System** | $dx/dt = rx(x+1) - x = x[rx + (r-1)]$ |
| **Equilibria** | $x^* = 0$ and $x^* = (1-r)/r$ |
| **Critical threshold** | $r_c = 1$ (equilibria collide at $x = 0$) |
| **Bifurcation type** | **Transcritical** |
| **$r < 1$** | $x^*=0$ is stable; $x^*=(1-r)/r > 0$ is unstable |
| **$r > 1$** | $x^*=0$ is unstable; $x^*=(1-r)/r < 0$ is stable |

A transcritical bifurcation is characterized by two equilibrium branches crossing at the critical parameter value and **exchanging stability** — neither branch is created or destroyed, they simply pass through each other. This is distinct from a saddle-node bifurcation (where two equilibria collide and annihilate) or a pitchfork bifurcation (where one equilibrium splits into three).
