## Question 4 — Stability of Each Equilibrium Point

For a **discrete-time map**, stability is determined by eigenvalue magnitude:

- $|\lambda| < 1$ → **stable** (nearby orbits converge)
- $|\lambda| > 1$ → **unstable** (nearby orbits diverge)
- $\lambda = 0$ → **super-stable** (collapse in finite steps)

*(For continuous ODEs the criterion is $\text{Re}(\lambda) < 0$; here it is $|\lambda| < 1$.)*

| Equilibrium | Eigenvalues | Classification |
|:------------|:-----------:|:--------------|
| $(1,\; 0,\; 0)$ — X wins | $0,\; 0$ | **Stable attractor** (super-stable) |
| $(0,\; 1,\; 0)$ — Y wins | $0,\; 0$ | **Stable attractor** (super-stable) |
| $(0,\; 0,\; 1)$ — Z wins | $0,\; 0$ | **Stable attractor** (super-stable) |
| $(\tfrac{1}{3},\tfrac{1}{3},\tfrac{1}{3})$ — Equal split | $2,\; 2$ | **Unstable** (repeller) |
| $(\tfrac{1}{2},\tfrac{1}{2},0)$ — X–Y tie | $0,\; 2$ | **Unstable saddle** |
| $(\tfrac{1}{2},0,\tfrac{1}{2})$ — X–Z tie | $0,\; 2$ | **Unstable saddle** |
| $(0,\tfrac{1}{2},\tfrac{1}{2})$ — Y–Z tie | $0,\; 2$ | **Unstable saddle** |

**For the specific case $(p_X, p_Y, p_Z) = (0.4, 0.3, 0.3)$:**  
$p_X$ is the largest, so the system converges to the X-wins equilibrium $(1, 0, 0)$.
Both eigenvalues equal zero — this is a **super-stable attractor**.
The Jacobian maps every nearby perturbation to zero in a single linear step,
meaning X's lead is self-reinforcing and the outcome is locked in quickly.

### General discussion

**Corner states (equilibria 1–3)** are the only truly stable outcomes.
All eigenvalues are zero, so every nearby initial condition converges
rapidly to whichever option started with the plurality — winner-takes-all.

**Equal split (equilibrium 4)** has both eigenvalues equal to $2 > 1$:
a pure repeller. Any slight imbalance is doubled each step and the
system moves away toward one of the corners.

**Two-option ties (equilibria 5–7)** are **saddle points** with $\{0, 2\}$.
The $\lambda=0$ direction is stable (the third option is drained to zero);
the $\lambda=2$ direction is unstable (any asymmetry between the tied pair
doubles each step until one wins).
