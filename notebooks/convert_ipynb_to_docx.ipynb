```python
from pathlib import Path
import subprocess

raw = input("Notebook path (or filename inside ./notebooks): ").strip()
ipynb_path = Path(raw)
if not ipynb_path.exists():
    alt = Path("notebooks") / raw
    if alt.exists():
        ipynb_path = alt
    else:
        raise FileNotFoundError(f"Not found: {raw}")

out_dir = Path("doc")
out_dir.mkdir(parents=True, exist_ok=True)
docx_path = out_dir / f"{ipynb_path.stem}.docx"

subprocess.run(["pandoc", str(ipynb_path), "-o", str(docx_path)], check=True)
print(f"Saved: {docx_path.resolve()}")
```
