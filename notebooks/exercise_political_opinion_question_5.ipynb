**Specific case:** $p_X = 0.4$,\; $p_Y = 0.3$,\; $p_Z = 0.3$

where $p_X$ = Republican, $p_Y$ = Democrat, $p_Z = 1 - p_X - p_Y$ = Independent.
All analysis below uses this starting state.


## Question 5 — Why Is an All-Z (Independent) State Not Feasible?

The model shows that $(p_X, p_Y, p_Z) = (0, 0, 1)$ is a mathematically stable
equilibrium (eigenvalues both zero). Starting from the specific case
$(0.4, 0.3, 0.3)$, $p_Z = 0.3$ is the smallest, so the system moves *away*
from the Z-wins state — $p_X$ takes over. Even so, the Z-wins basin exists
for initial conditions where $p_Z > p_X$ and $p_Z > p_Y$.

In practice, however, an all-Independent outcome is not feasible:

1. **Independents are not a coherent bloc.** Being "Independent" means absence
   of party affiliation, not a shared platform. In real elections, self-identified
   Independents split between X and Y candidates and do not converge on a single
   Independent option.

2. **No governing majority.** A 100% Independent electorate has no unified agenda.
   Governance requires an organized majority, which a purely Independent population
   cannot provide.

3. **Structural barriers.** Winner-take-all elections and ballot-access rules
   structurally favor the two established parties — features the model ignores.

4. **Model homogeneity assumption.** The model treats all $p_Z$ voters as
   equivalent and equally responsive. In reality, voter identities are heterogeneous
   and partisan affiliations are sticky.

The all-Z state is a mathematical artifact of the model's symmetry;
real-world factors make it politically infeasible.
