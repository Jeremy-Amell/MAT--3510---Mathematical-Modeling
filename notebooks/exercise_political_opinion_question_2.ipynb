## Question 2 — Jacobian Matrix at Each Equilibrium

The system is a 2-D discrete map $F(p_X, p_Y) = (p_X + \Delta p_X,\; p_Y + \Delta p_Y)$
with $p_Z = 1 - p_X - p_Y$.  The Jacobian is:

$$J = DF\big|_{\text{eq}} = I + \frac{\partial(\Delta p_X,\, \Delta p_Y)}{\partial(p_X,\, p_Y)}\bigg|_{\text{eq}}$$

Because the update rule contains $\max$ terms, it is **piecewise linear**.
The Jacobian is evaluated analytically in the dominant region near each equilibrium.

The specific case $(p_X, p_Y, p_Z) = (0.4, 0.3, 0.3)$ converges to **equilibrium 1**
$(p_X, p_Y, p_Z) = (1, 0, 0)$, so **J1** is the relevant Jacobian for this case.


```python
import numpy as np

# Analytical Jacobians — variable ordering: (p_X, p_Y); p_Z = 1 - p_X - p_Y

J1 = np.array([[0., 0.], [0., 0.]])          # (1,   0,   0)  X wins  <-- specific case
J2 = np.array([[0., 0.], [0., 0.]])          # (0,   1,   0)  Y wins
J3 = np.array([[0., 0.], [0., 0.]])          # (0,   0,   1)  Z wins
J4 = np.array([[2., 0.], [0., 2.]])          # (1/3,1/3,1/3) Equal split
J5 = np.array([[ 1.,-1.], [-1., 1.]])        # (1/2,1/2, 0)  X-Y tie
J6 = np.array([[ 2., 1.], [ 0., 0.]])        # (1/2, 0, 1/2) X-Z tie
J7 = np.array([[ 0., 0.], [ 1., 2.]])        # (0, 1/2,1/2)  Y-Z tie

labels = [
    '(1,   0,   0)   X wins  [specific case]',
    '(0,   1,   0)   Y wins',
    '(0,   0,   1)   Z wins',
    '(1/3,1/3, 1/3)  Equal split',
    '(1/2,1/2,  0)   X-Y tie',
    '(1/2, 0,  1/2)  X-Z tie',
    '(0,  1/2, 1/2)  Y-Z tie',
]

for name, J in zip(labels, [J1,J2,J3,J4,J5,J6,J7]):
    print(f'Equilibrium {name}:')
    print(f'  J = | {J[0,0]:+5.1f}  {J[0,1]:+5.1f} |')
    print(f'      | {J[1,0]:+5.1f}  {J[1,1]:+5.1f} |')
    print()
```
